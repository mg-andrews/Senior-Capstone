```python
#This script aims to develop a csv file of two columns headered as "t" and "mmHg" which model the inflation and deflation of a BP cuff
#Author: Joshua Van Tuyl
#VitalCentric Project

#Imports
import numpy as np
import pandas as pd
import math
import matplotlib.pyplot as plt

#Importing healthy blood pressure data
b = pd.read_csv("Healthy_BP_JVT_12_2025.csv")
#Saving csv columns as data lists
healthy_data = b.to_dict(orient="list")
#Saving each column as a discrete list
time = healthy_data["0"]
healthy_bp = healthy_data["80"]
#Healthy BP data called h(t)
```

```python
#Create Gaussian Function g(t)
#Takes in inputs of time and outputs values
def gauss(t, mu, sigma):
    g = np.exp(-1 * ((t - mu)**2) / (2 * sigma**2))
    return g
```

```python
#Verifying Gaussian Function
time = []
gaussian_pressure = []

#variable mu determines the peak of the gaussian curve, so it must be the systolic pressure
mu = 90 
#sigma determines the flatness or sharpness of the curve, should be the upper and lower ends of effective measurement range
sigma = 20
#Assuming symmetry for now

for x in range(0, len(b)):
    time.append(x)
    gaussian_pressure.append(gauss(x, mu, sigma))
```

```python
#Convoluting between g(t) and h(t) to create modulated dataset m(t)
m = np.convolve(gaussian_pressure, healthy_bp, mode="full")
```

```python
plt.plot(m)
```
